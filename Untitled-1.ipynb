```python
import os
import glob
from langchain_community.document_loaders import UnstructuredMarkdownLoader

# ✅ Set the actual path to your Markdown files
knowledge_directory = r"C:\Users\youss\Downloads\threat_groups_md"

# Step 1: Locate all Markdown files
group_files = glob.glob(os.path.join(knowledge_directory, "*.md"))

# Step 2: Initialize container
md_docs = []

print("[+] Loading Group markdown files..")

# Step 3: Load each Markdown file
for group_path in group_files:
    if os.path.getsize(group_path) == 0:
        print(f" [!] Skipping empty file: {os.path.basename(group_path)}")
        continue

    try:
        loader = UnstructuredMarkdownLoader(group_path)
        docs = loader.load()
        if docs:
            md_docs.extend(docs)
            print(f" [*] Loaded: {os.path.basename(group_path)} ({len(docs)} docs)")
        else:
            print(f" [!] No content extracted from: {os.path.basename(group_path)}")
    except Exception as e:
        print(f" [!] Error loading {os.path.basename(group_path)}: {e}")

# Step 4: Summary
print(f'[+] Number of .md documents processed: {len(md_docs)}')

# Step 5: Safely preview one of the pages
if len(md_docs) > 5:
    print(md_docs[5].page_content)
elif md_docs:
    print("[i] Fewer than 6 docs found. Showing first one:")
    print(md_docs[0].page_content)
else:
    print("[!] No markdown documents were loaded. Check folder or file format.")
```

```python
from llama_cpp import Llama
from mitreattack.stix20 import MitreAttackData

# Load the MITRE ATT&CK dataset
ENTERPRISE_ATTACK_PATH = r"C:\Users\youss\Downloads\enterprise-attack (1).json"
attack_data = MitreAttackData(ENTERPRISE_ATTACK_PATH)

# Extract groups
groups = attack_data.get_groups()

# Use description of the first group (or more)
if not groups:
    print("❌ No groups found in the dataset.")
else:
    # Extract text (e.g. from the first group's description)
    text = groups[0]["description"]

    # ✅ Load the LLaMA model
    LLM_MODEL_PATH = r"C:/Users/youss/Downloads/llama-2-7b.Q4_K_M.gguf"
    llm = Llama(model_path=LLM_MODEL_PATH, n_ctx=2048)

    # ✅ Tokenize the text
    tokens = llm.tokenize(text.encode("utf-8"))
    print(f"🧠 Token count: {len(tokens)}")
    print(f"🔢 Tokens: {tokens[:50]}{'...' if len(tokens) > 50 else ''}")
```

```python
# === Step 3: Tokenization Function Using llama-cpp ===
def llama_token_len(text):
    return len(llm.tokenize(text.encode("utf-8")))

# === Step 4: Compute Token Counts (Fix: use .description instead of ["description"]) ===
token_counts = [llama_token_len(group.description) for group in groups if hasattr(group, "description")]

# === Step 5: Print Token Statistics ===
print(f"""[+] Token Counts (Group Descriptions):
Min: {min(token_counts)}  # Minimum tokens
Avg: {int(sum(token_counts) / len(token_counts))}  # Average tokens
Max: {max(token_counts)}  # Maximum tokens
""")
```

```python
from langchain.text_splitter import RecursiveCharacterTextSplitter

# Token length function for LLaMA tokenizer
def llama_token_len(text):
    return len(llm.tokenize(text.encode("utf-8")))

# Print a message indicating the initialization of RecursiveCharacterTextSplitter
print('[+] Initializing RecursiveCharacterTextSplitter..')

# Create an instance of RecursiveCharacterTextSplitter
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,           # Max tokens per chunk
    chunk_overlap=50,         # Overlap tokens between chunks
    length_function=llama_token_len,  # Use LLaMA-based token count
    separators=['\n\n', '\n', ' ', '']  # Hierarchical splitting rules
)
```

```python
group_texts = [group.description for group in groups if hasattr(group, 'description')]
chunks = text_splitter.create_documents(group_texts)
print(f"[+] Created {len(chunks)} chunks.")
```

```python
from sentence_transformers import SentenceTransformer
from langchain.vectorstores import FAISS
from langchain.embeddings import HuggingFaceEmbeddings

# Load a local embedding model
print("[+] Loading local embedding model...")
embedding_model = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")

# Embed and store in FAISS
print("[+] Creating vector store with FAISS...")
db = FAISS.from_documents(chunks, embedding_model)
```

```python
retriever = db.as_retriever(search_kwargs={"k":5})
```

```python
query = "How do cyber threat actors gain initial access to target systems?"
```
